## **Padding and Strides**
- When we are covolving the filter on the image, we have the output reduced. 
- This is the problem with covolution i.e at each layer we have the image size keep getting reduced, there is loss of information
- The boundary pixels are contributing low the output and the middle pixels are getting covolved more thus boundary pixels are given less imporatance. 


## **PADDING**
- We add two rows and two columns in the image. 
- The output we create is called padding
- The each padded pixels have zero value callued zero padding.
- Thus the boundary values are convolved multiple times.
- After the paddin the formula changes to (n+2p-f+1) from (n-f+1) where p is the padding option.

```python
import tensorflow
from tensorflow import keras
from keras.layers import Dense,Conv2D,Flatten
from keras import Sequential
from keras.datasets import mnist
```

```python
(X_train,y_train),(X_test,y_test)=mnist.load_data()
```

```python
model=Sequential()
model.add(Conv2D(32,kernel_size=(3,3),padding='valid',activation='relu',input_shape=(28,28,1)))
'''Conv2d is for cnn, 32 is the number of the filtes with 
filter size (3,3), valid padding means no padding used'''
model.add(Conv2D(32,kernel_size=(3,3),padding='valid',activation='relu'))
model.add(Conv2D(32,kernel_size=(3,3),padding='valid',activation='relu'))
'''we have used the three convolution
then we pass the model to flatten'''
model.add(Flatten())
model.add(Dense(128,activation='relu'))
model.add(Dense(10,activation='softmax'))
```

```python
model.summary()#gives the summary of the model
```

```python
model=Sequential()
model.add(Conv2D(32,kernel_size=(3,3),padding='same',activation='relu',input_shape=(28,28,1)))
'''Conv2d is for cnn, 32 is the number of the filtes with 
filter size (3,3), same padding means no padding used'''
model.add(Conv2D(32,kernel_size=(3,3),padding='same',activation='relu'))
model.add(Conv2D(32,kernel_size=(3,3),padding='same',activation='relu'))
'''we have used the three convolution
then we pass the model to flatten'''
model.add(Flatten())
model.add(Dense(128,activation='relu'))
model.add(Dense(10,activation='softmax'))
```

## **STRIDES**
- When we previously learned the convolution we had the shift of the filter by one step. Thus we can say that stride was (1x1).
- We can also put stride as (2x2). Here we move by two steps. 
- **What is the impact of strides on the result?**
    - The first thing is we will have the change in the size of the feature map. 
    - Then the formula becoms $\frac{n-f}{s}+1$
    - if we also included the padding $\frac{n+2p-f}{2}+1$
- The increase in the stride is responsible for information loss
- If we use the stride other than (1x1) we call it strided feature map.

- Suppose we have stiride (2x2) and the image(7x6) with filter (3x3), then we cannot perform the covolution on the last row. 
- Here in such opeartion we may get the value of the feature map in point thus we take the floor value for the decimal. This means we are lacking for the number of the pixels for the given stride.

### **WHAT IS THE IMPORTANCE OF STRIDES**
- When only the high level features are desired from the image. Then low level features are not caputured.
- Computing: For faster training, increase the value of the strides. 

```python
model=Sequential()
model.add(Conv2D(32,kernel_size=(3,3),strides=(2,2),padding='same',activation='relu',input_shape=(28,28,1)))
'''Conv2d is for cnn, 32 is the number of the filtes with 
filter size (3,3), same padding means no padding used'''
model.add(Conv2D(32,kernel_size=(3,3),strides=(2,2),padding='same',activation='relu'))
model.add(Conv2D(32,kernel_size=(3,3),strides=(2,2),padding='same',activation='relu'))
'''we have used the three convolution
then we pass the model to flatten'''
model.add(Flatten())
model.add(Dense(128,activation='relu'))
model.add(Dense(10,activation='softmax'))
```

```python
model.summary()
```
