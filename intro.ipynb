### **What is CNN**

- Convolution neural network, also known as convnet, or CNNs, are a special kind of neural network for processing data that has a known grid-like topology like time series data(1D) or images(2D).
- Here it uses the convolution opeartion.


**The layers present in the CNN:**
- Convolution
- Pooling
- FC layer

- It is inspired by the visual cortex.
