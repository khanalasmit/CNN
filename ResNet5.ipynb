```python
#import image from keras preprocessing
from tensorflow.keras.preprocessing import image
#import preprocess_input from tensorflow keras applications resnet50
from tensorflow.keras.applications.resnet50 import preprocess_input
import numpy as np
img_path='dog.webp'
```

```python
img=image.load_img(img_path,target_size=(224,224))
img=image.img_to_array(img)#turn image into array
#expand the dimensions so that its understood by our network 
#img.shape truns from (224,224,3) into (1,224,224,3)
img=np.expand_dims(img,axis=0)
img=preprocess_input(img)#pre-process the image in the same way training images were
```

```python
#import resnet50 and decode_predictions from tensorflow.keras.application.resnet50
from tensorflow.keras.applications.resnet50 import ResNet50, decode_predictions
#Instantiate a ResNEt50 model with imagenet weights
model=ResNet50(weights='imagenet')

preds=model.predict(img)

print('Predicted:',decode_predictions(preds,top=2)[0])
```
