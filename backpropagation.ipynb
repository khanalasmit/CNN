### **Backpropagation in CNN**
________________________________
**Trainable Parameters**
- weights(inputs to the internal neurons)
- bias
- Loss function is: $L=-Y_ilog(\hat{Y_i})-(1-Y_i)log(1-\hat{Y_i})$


**Forward propagation**
- $Z_1=Conv(x,w_1)+b_1$
- $A_1=relu(z_1)$
- $P_1=maxpool(A_1)$
- $F=Flatten(P_1)$
- $z_2=w_2F+b_2$
- $A_2=\sigma(z_2)$
